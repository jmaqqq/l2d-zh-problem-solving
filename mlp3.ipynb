```python
import torch
import torch.nn.functional as F
from torch import nn

class CenteredLayer(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self,x):
        return (x-x.mean()).sum()

layer = CenteredLayer()
x = torch.arange(6, dtype=torch.float32)
layer(x)
```

```python
net=nn.Sequential(
    nn.Linear(8,128),
    CenteredLayer()
)
y=net(torch.rand(4,8))
print(y)
```

```python
class MyLinear(nn.Module):
    def __init__(self,in_units,units):
        super().__init__()
        self.weight=nn.Parameter(torch.randn(in_units,units))
        self.bias=nn.Parameter(torch.randn(units))

    def forward(self,x):
        linear=torch.matmul(x,self.weight.data)+self.bias.data
        return F.relu(linear)

dense=MyLinear(5,3)
dense.weight
```

```python
y=dense(torch.arange(10.).reshape(2, 5))
print(y)
print(y.shape)
```

```python
net=nn.Sequential(
    MyLinear(64,8),
    MyLinear(8,1)
)
net(torch.rand(2,64))
```
