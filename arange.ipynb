
当张量的形状符合特定规则时，PyTorch 可以自动扩展较小张量的形状，使两个张量能够逐元素运算。

- 如果两个张量的维度数量不同，就在维度较少的形状前面补 `1`。
- 从最后一维开始比较：每一维的大小必须相等，或者其中一个为 `1`，才能广播。

| 张量 A | 张量 B | 能否广播相加 | 结果形状 |
|---|---|---|---|
| `(3, 1)` | `(3, 2)` | 可以 | `(3, 2)` |
| `(3, 1)` | `(1, 3, 2)` | 可以 | `(1, 3, 2)` |
| `(3, 2)` | `(3, 4)` | 不可以 | 形状不兼容 |


```python
a2=torch.tensor([1,4,7]).reshape(3,1)
y3=torch.arange(6,dtype=torch.float32).reshape((1,3,2))
a2,a2.shape,y3,y3.shape,a2+y3
```

```python
x=torch.tensor([3])
before=id(x)
x=x+x
id(x)==before
```

- 原地操作：修改原有张量的元素，Z 仍然引用原来的对象。
- 普通赋值：通常会计算出一个新的张量，再让变量 Z 引用这个新张量。原来的张量对象不会因此被修改。

```python
x=torch.tensor([3])
before=id(x)
x[:]=x+x
id(x)==before
```

```python
x=torch.tensor([3])
before=id(x)
x+=x
id(x)==before
```
