```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
def vgg_block(num_convs,in_channels,out_channels):
    layers=[]
    for _ in range(num_convs):
        layers.append(nn.Conv2d(in_channels,out_channels,kernel_size=3,padding=1))
        layers.append(nn.ReLU())
        in_channels=out_channels
    layers.append(nn.MaxPool2d(kernel_size=2,stride=2))
    return nn.Sequential(*layers)
```

```python
conv_arch = ((1, 64), (1, 128), (2, 256), (2, 512), (2, 512))
def vgg1(conv_arch):
    conv_blks=[]
    in_channels=1
    for (num_convs,out_channels) in conv_arch:
        block=vgg_block(num_convs,in_channels,out_channels)
        conv_blks.append(block)
        in_channels=out_channels
    return nn.Sequential(*conv_blks)
```

```python
line_arch=((1,4096),(1,4096),(1,10))
def vgg2(line_arch,in_features=512*7*7,drop=0.5):
    line_blks=[]
    for(num_linears,out_linears) in line_arch:
        for _ in range(num_linears):
            line_blks.append(nn.Linear(in_features,out_linears))
            in_features=out_linears
            is_last=((num_linears,out_linears)==line_arch[-1] and _ ==num_linears-1)
            if not is_last:
                line_blks.append(nn.ReLU())
                line_blks.append(nn.Dropout(drop))
    return nn.Sequential(*line_blks)
```

```python
alexnet=nn.Sequential(
    vgg1(conv_arch),
    nn.Flatten(),
    vgg2(line_arch),    
    )
```
