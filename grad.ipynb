````python
import torch
````

````python
x=torch.arange(4.0)
x
````

````python
x.requires_grad_(True)
x.grad
````

````python
x=torch.arange(4.0,requires_grad=True)
x
````

````python
y=2*torch.dot(x,x)
y
````

````python
y.backward()
x.grad
````

````python
x.grad==4*x
````

````python
x.grad.zero_()
y=x.sum()
y.backward()
x.grad
````

如果没有x.grad.zero_()梯度会累加
```
y=x.sum()
y.backward()
x.grad
```
最后值可能是tensor([2,2,2,2])

````python
x=torch.arange(4.0,requires_grad=True)
y=x*x
y.backward(torch.ones_like(y))
x.grad
````

````python
x=torch.arange(4.0,requires_grad=True)
y=x*x
y.sum().backward()
x.grad
````

````python
x.grad.zero_()
x=torch.arange(4.0,requires_grad=True)
y=x*x
u=y.detach()
z=u*x
z.sum().backward()
x.grad==u 
````

````python
x.grad.zero_()
y.sum().backward()
x.grad==2*x
````

````python
x=torch.arange(4.0,requires_grad=True)
y=x*x

z=y*x
z.sum().backward()
x.grad==y
````

````python
def f(a):
    b=a*2
    while b.norm()<1000:
        b=b*2
        if b.sum()>0:
             c=b
        else:
            c=100*b
    return c   
````

````python
a=torch.randn(size=(),requires_grad=True)
d=f(a)
d.backward()
a.grad==d/a
````

````python

a = torch.tensor(0.0, requires_grad=True)
d=f(a)
d.backward()
a.grad==d/a
````
